```python
# 3. Check Missing Values
# ==========================================

# Count missing values in each column
missing_values = df.isnull().sum()

# Print missing values
print("\nMissing Values in Each Column:")
print(missing_values)

```

```python
# 4. Remove Irrelevant Columns (if any)
# ==========================================

# Drop unnecessary columns like index or unique ID
df.drop(['Unnamed: 0', 'key'], axis=1, inplace=True, errors='ignore')
```

```python
# 5. Handle Missing Values
# ==========================================

# Strategy 1: Drop rows with missing values
# (Suitable when missing values are small in number)
df = df.dropna()
```

```python
# 6. Verify Missing Values After Handling
# ==========================================

print("\nMissing Values After Handling:")
print(df.isnull().sum())
```

```python
# 7. Basic Validation
# ==========================================

# Remove invalid values (extra cleaning)
df = df[df['fare_amount'] > 0]          # Remove negative fares
df = df[df['passenger_count'] > 0]      # Remove invalid passenger counts

# Display cleaned dataset info
print("\nDataset Info After Cleaning:")
print(df.info())
```

```python
# 11. Verify Dataset Shape After Cleaning
# ==========================================

# Print updated dataset shape
print("\nDataset Shape After Handling Missing Data:")
print(df.shape)
```

```python
# 3. Convert Datetime Column
# ==========================================

# Convert pickup_datetime into proper datetime format
df['pickup_datetime'] = pd.to_datetime(df['pickup_datetime'])

```

```python
# Remove extreme fare values (outliers)
df = df[df['fare_amount'] < 100]
```

```python
# Remove invalid passenger count
df = df[(df['passenger_count'] >= 0) & (df['passenger_count'] <= 6)]
```

```python
# Remove invalid latitude/longitude values
df = df[
    (df['pickup_latitude'].between(40, 42)) &
    (df['pickup_longitude'].between(-75, -72)) &
    (df['dropoff_latitude'].between(40, 42)) &
    (df['dropoff_longitude'].between(-75, -72))
]
```

```python
#remove fare outliers
Q1 = df['fare_amount'].quantile(0.25)
Q3 = df['fare_amount'].quantile(0.75)
IQR = Q3 - Q1

df = df[(df['fare_amount'] >= Q1 - 1.5*IQR) &
        (df['fare_amount'] <= Q3 + 1.5*IQR)]
```

```python
# Extract useful time-based features
df['hour'] = df['pickup_datetime'].dt.hour       # Hour of ride
df['day'] = df['pickup_datetime'].dt.day         # Day of month
df['month'] = df['pickup_datetime'].dt.month     # Month
df['weekday'] = df['pickup_datetime'].dt.weekday # Day of week (0 = Monday)
# Peak hours feature
df['is_peak'] = df['hour'].apply(lambda x: 1 if (7 <= x <= 10) or (17 <= x <= 20) else 0)
df['is_night'] = df['hour'].apply(lambda x: 1 if (0 <= x <= 5) else 0)

# Weekend feature (correct logic: 5=Saturday, 6=Sunday if using weekday)
df['is_weekend'] = df['day'].apply(lambda x: 1 if x in [5, 6] else 0)

```

```python
# 4. Create Time of Day Feature
# ==========================================

# Function to classify time of day
def get_time_of_day(hour):
    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"

# Apply function
df['time_of_day'] = df['hour'].apply(get_time_of_day)

```

```python
# 5. Calculate Distance (Haversine Formula)
# ==========================================

def haversine(lat1, lon1, lat2, lon2):
    R = 6371  # Earth radius in km

    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2)**2
    c = 2 * np.arcsin(np.sqrt(a))

    return R * c
```

```python
# Create distance feature
df['distance_km'] = haversine(
    df['pickup_latitude'], df['pickup_longitude'],
    df['dropoff_latitude'], df['dropoff_longitude']
)

```

```python
# Remove zero or unrealistic distances
df = df[df['distance_km'] > 0]
df = df[df['distance_km'] < 30]   # remove unrealistic long trips
```

```python
# Remove distance outliers using IQR
Q1 = df['distance_km'].quantile(0.25)
Q3 = df['distance_km'].quantile(0.75)
IQR = Q3 - Q1

df = df[(df['distance_km'] >= Q1 - 1.5*IQR) &
        (df['distance_km'] <= Q3 + 1.5*IQR)]
```

```python
# 6. Create Distance Category Feature
# ==========================================

def distance_category(dist):
    if dist < 2:
        return "Short"
    elif dist < 10:
        return "Medium"
    else:
        return "Long"

df['distance_category'] = df['distance_km'].apply(distance_category)
```

```python
# 7. Encode Categorical Features
# ==========================================

# Convert categorical features into numeric using one-hot encoding
df = pd.get_dummies(df, columns=['time_of_day', 'distance_category'], drop_first=True)

```

```python
# 8. Final Dataset Preview
# ==========================================

print("Final Dataset with New Features:")
print(df.head())

print("\nUpdated Columns:")
print(df.columns)
```

```python
#task: Split the dataset

# 1. Import Required Libraries
# ==========================================

from sklearn.model_selection import train_test_split
```

```python
# 2. Define Features (X) and Target (y)
# ==========================================

# Distance interaction feature (re-added as it was deleted earlier)
df['distance_hour_interaction'] = df['distance_km'] * df['hour']

# X = Independent variables (inputs)
X = df[['distance_km',
        'hour',
        'is_peak',
        'is_night',
        'is_weekend',
        'distance_hour_interaction']]
# y = Dependent variable (target)
y = df['fare_amount']
```

```python
# 3. Split the Dataset
# ==========================================

# Split data into training (80%) and testing (20%)
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,      # 20% data for testing
    random_state=42     # Ensures same split every time
)
```

```python
# 4. Check the Shape of Split Data
# ==========================================

print("Training Features Shape:", X_train.shape)
print("Testing Features Shape:", X_test.shape)
print("Training Target Shape:", y_train.shape)
print("Testing Target Shape:", y_test.shape)
```

```python
#data scaling/normalization

# 1. Import Required Library
# ==========================================

from sklearn.preprocessing import StandardScaler
```

```python
num_features = ['distance_km', 'hour', 'distance_hour_interaction']
cat_features = ['is_peak', 'is_night', 'is_weekend']
```

```python
# 2. Initialize Scaler
# ==========================================

scaler = StandardScaler()
X_train_num = scaler.fit_transform(X_train[num_features])
X_test_num = scaler.transform(X_test[num_features])
```

```python
import numpy as np

X_train_scaled = np.concatenate([X_train_num, X_train[cat_features].values], axis=1)
X_test_scaled = np.concatenate([X_test_num, X_test[cat_features].values], axis=1)
```

```python
# 3. Apply Scaling
# ==========================================

# Fit scaler ONLY on training data and transform both sets
X_train_scaled = scaler.fit_transform(X_train)

# Use same scaler on test data (important!)
X_test_scaled = scaler.transform(X_test)
```

```python
# 4. Check Results
# ==========================================

print("Before Scaling (First Row):")
print(X_train.iloc[0])

print("\nAfter Scaling (First Row):")
print(X_train_scaled[0])
```

```python
#task: select and train regression model

# 1. Import Required Libraries
# ==========================================

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
```

```python
# 2. Initialize Models
# ==========================================

# Linear Regression Model
lr_model = LinearRegression()

# Random Forest Model
rf_model = RandomForestRegressor(
    n_estimators=200,      # more trees = better learning
    max_depth=20,          # control overfitting
    min_samples_split=5,  # minimum samples to split a node
    min_samples_leaf=2,    # minimum samples at leaf node
    random_state=42
)
```

```python
import numpy as np

y_train_log = np.log1p(y_train)
y_test_log = np.log1p(y_test)

lr_model.fit(X_train_scaled, y_train_log)

y_pred_log = lr_model.predict(X_test_scaled)
y_pred = np.expm1(y_pred_log)
```

```python
# 3. Train Models
# ==========================================

# Train Linear Regression
lr_model.fit(X_train_scaled, y_train)

# Train Random Forest
rf_model.fit(X_train, y_train)
```

```python
# 4. Make Predictions
# ==========================================

# Predictions using Linear Regression
y_pred_lr = lr_model.predict(X_test_scaled)

# Predictions using Random Forest
y_pred_rf = rf_model.predict(X_test)

print("Models trained successfully!")
```

```python
#task : evaluate the regression model

# 1. Import Evaluation Metrics
# ==========================================

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

```python
# 2. Evaluate Linear Regression
# ==========================================


mae_lr = mean_absolute_error(y_test, y_pred_lr)
mse_lr = mean_squared_error(y_test, y_pred_lr)
rmse_lr = np.sqrt(mse_lr)
r2_lr = r2_score(y_test, y_pred_lr)

print("Linear Regression Performance:")
print("MAE:", mae_lr)
print("MSE:", mse_lr)
print("RMSE:", rmse_lr)
print("R2 Score:", r2_lr)
```

```python
# 3. Evaluate Random Forest
# ==========================================

mae_rf = mean_absolute_error(y_test, y_pred_rf)
mse_rf = mean_squared_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mse_rf)
r2_rf = r2_score(y_test, y_pred_rf)

print("\nRandom Forest Performance:")
print("MAE:", mae_rf)
print("MSE:", mse_rf)
print("RMSE:", rmse_rf)
print("R2 Score:", r2_rf)
```

```python
#task: fine tuning the model


# 1. Import Grid Search
# ==========================================

from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import RandomizedSearchCV
```

```python
# 2. Define Model
# ==========================================

rf = RandomForestRegressor(random_state=42)
```

```python
# 3. Define Hyperparameter Grid
# ==========================================

param_dist = {
    'n_estimators': [200, 300, 400],
    'max_depth': [10, 20, 30, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

```

```python
# 4. Apply random Search
# ==========================================

random_search = RandomizedSearchCV(
    rf,
    param_distributions=param_dist,
    n_iter=10,         # tries 10 combinations
    cv=3,
    scoring='r2',
    random_state=42,
    n_jobs=-1
)

# Train random Search
random_search.fit(X_train, y_train)
```

```python
# 5. Best Parameters
# ==========================================

print("Best Parameters:", random_search.best_params_)
```

```python
# 6. Use Best Model
# ==========================================


best_rf = random_search.best_estimator_

# Predict using best model
y_pred_best = best_rf.predict(X_test)
```
